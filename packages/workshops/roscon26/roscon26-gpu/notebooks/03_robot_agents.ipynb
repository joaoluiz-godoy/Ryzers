# Robot Agents with RAI

![](images/manipulation_demo.gif)

RAI (Robot Agent Interface) is an open-source framework from Robotec.ai for building and deploying Embodied AI agents on robots. It connects a Large Language Model to a robot's ROS 2 stack, letting the model perceive the scene, reason about a natural-language instruction, and call the robot's tools to carry it out. We can connect it to Lemonade to run the model locally.

## Goals

* Learn how to use AI Agents for Physical AI

The first step will be to open a terminal session.

## RAI Setup

Image of opening terminal on IDE: ![](images/new_terminal.png)

### Enter container
```
ryzers run bash
```

### Point RAI at your local model
```
source lemonade_env.sh
```

### Launch the demo (headless)
Renders the simulation without a monitor and streams it to your browser next to the chat — no local display needed.
```
bash /ryzers/workshop/manipulation_demo_headless.sh
```

The O3DE simulation can take a minute or two to load — wait until the objects spawn on the table. Then open [http://localhost:8501](http://localhost:8501): the **left panel is the live simulation** and the **right panel is the robot-agent chat**, all on a single page.

#### (Optional) - Use a different local model
```
source lemonade_env.sh <model-name>
```

## Suggested Prompts
Once the scene has loaded, use the chat panel on the right to command the robot agent — each instruction is reasoned about and executed by your local Gemma model on the iGPU, nothing leaving the machine:

- **Stack objects:** "Stack the cubes on top of each other."

- **Sort by color:** "Group the cubes by color."

- **Relocate:** "Move the red cube to the left side of the table."

- **Reason first:** "What objects can you see on the table, and where are they?"
